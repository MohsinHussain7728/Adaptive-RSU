```python
import matplotlib.pyplot as plt

vehicle_density = [40, 60, 80]
pdr_with_uav = [88.5, 84.1, 79.3]

plt.figure(figsize=(8, 5))
plt.plot(vehicle_density, pdr_with_uav, marker='o', linestyle='-', color='#2ecc71', linewidth=2, markersize=8, label='UAV Relay (AODV)')

plt.title('Scalability Analysis: PDR under High Congestion', fontweight='bold', fontsize=14)
plt.xlabel('Number of Vehicles (Density)', fontweight='bold', fontsize=12)
plt.ylabel('Packet Delivery Ratio (%)', fontweight='bold', fontsize=12)
plt.ylim(50, 100)
plt.xticks(vehicle_density)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(loc='lower left')

for x, y in zip(vehicle_density, pdr_with_uav):
    plt.text(x, y + 2, f'{y}%', ha='center', fontweight='bold')

plt.tight_layout()
plt.savefig('scalability_graph_high_density.png', dpi=300)
print("Success! Graph saved as 'scalability_graph_high_density.png'")
```
